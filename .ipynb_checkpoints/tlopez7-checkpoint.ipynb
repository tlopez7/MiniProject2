```python
from woc.remote import WocMapsRemote
import pandas as pd

woc = WocMapsRemote(base_url="https://worldofcode.org/api/")

projects = ['gavinsimpson/permute']
list_df_commits = []
for prj in projects:
    prj = prj.lower().replace('/', '_', 2)
    commits = woc.get_values('p2c', prj)
    df = pd.DataFrame(commits, columns=["sha1"])
    df['project'] = prj
    list_df_commits.append(df)

df = pd.concat(list_df_commits)
df.to_csv('df_commits.csv', index=False)
print(df.head())
print("Total commits found:", len(df))
```

```python
from tqdm import tqdm
import time

chunks = [df['sha1'][x:x+10] for x in range(0, len(df), 10)]
commit_data = []

for chunk in tqdm(chunks):
    res, err = woc.get_values_many('commit.tch', chunk.to_list())
    res = {k: v[0] for k, v in res.items()}
    time.sleep(1)  # be polite to the API

    if err:
        print('Got Errors', err)

    for commit_sha, commit in res.items():
        commit_data.append({
            'commit': commit_sha,
            'author': commit[2][0],
            'author_time': int(commit[2][1]),
            'message': commit[4],
        })

df_commit_data = pd.DataFrame(commit_data)
df_commit_data['project'] = 'gavinsimpson_permute'
print(df_commit_data.head())
print("Total commits with details:", len(df_commit_data))
```

```python
dfinf = df_commit_data.rename(columns={
    'project': 'project_wocid',
    'commit': 'commit_sha1',
    'author_time': 'time',
    'message': 'commit message'
})[['project_wocid', 'commit_sha1', 'author', 'time', 'commit message']]

dfinf.to_csv('tlopez7_project_summary.csv', index=False, sep=';', mode='w', header=True)
print(dfinf.head())
```

```python
import time

def fetch_project_commits(github_name):
    prj = github_name.lower().replace('/', '_', 2)
    commits = woc.get_values('p2c', prj)
    sha1s = [c for c in commits]

    chunks = [sha1s[x:x+10] for x in range(0, len(sha1s), 10)]
    commit_data = []
    for chunk in tqdm(chunks, desc=prj):
        res, err = woc.get_values_many('commit.tch', chunk)
        res = {k: v[0] for k, v in res.items()}
        time.sleep(1)
        if err:
            print('Errors for', prj, ':', err)
        for sha, commit in res.items():
            commit_data.append({
                'project_wocid': prj,
                'commit_sha1': sha,
                'author': commit[2][0],
                'time': int(commit[2][1]),
                'commit message': commit[4],
            })
    return pd.DataFrame(commit_data)

remaining_projects = [
    'openclimatefix/Satip', 'JuliaOpt/Convex.jl', 'eljost/pysisyphus',
    'simonvh/genomepy', 'brentp/cyvcf2', 'DiCarloLab-Delft/PycQED_py3',
    'Benchmarking-Initiative/Benchmark-Models-PEtab', 'gsdlab/clafer',
    'Acellera/htmd'
]

for p in remaining_projects:
    df_p = fetch_project_commits(p)
    df_p.to_csv('tlopez7_project_summary.csv', index=False, sep=';', mode='a', header=False)
    print(f"{p}: {len(df_p)} commits saved")
```

```python
import requests, time
import pandas as pd
import requests, time

github_repos = [
    'gavinsimpson/permute', 'openclimatefix/Satip', 'JuliaOpt/Convex.jl',
    'eljost/pysisyphus', 'simonvh/genomepy', 'brentp/cyvcf2',
    'DiCarloLab-Delft/PycQED_py3', 'Benchmarking-Initiative/Benchmark-Models-PEtab',
    'gsdlab/clafer', 'Acellera/htmd'
]

github_stats = []
for repo in github_repos:
    r = requests.get(f'https://api.github.com/repos/{repo}')
    data = r.json()
    commits_r = requests.get(f'https://api.github.com/repos/{repo}/commits', params={'per_page': 1})
    last_commit_date = commits_r.json()[0]['commit']['committer']['date'] if commits_r.status_code == 200 else None
    github_stats.append({
        'project': repo,
        'stars': data.get('stargazers_count'),
        'forks': data.get('forks_count'),
        'last_commit_date': last_commit_date,
    })
    time.sleep(1)

df_github_stats = pd.DataFrame(github_stats)
print(df_github_stats)
```

```python
import pandas as pd
import requests, time
df_all = pd.read_csv('tlopez7_project_summary.csv', sep=';')

woc_stats = df_all.groupby('project_wocid').agg(
    num_commits=('commit_sha1', 'count'),
    num_authors=('author', 'nunique'),
    max_time=('time', 'max'),
    min_time=('time', 'min'),
).reset_index()

print(woc_stats)
```
